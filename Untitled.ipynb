```python
sq=lambda a:a*a
re=lambda l,b:l*b
tri=lambda l,h:0.5*l*h;
while True:
    print("\n 1.square \n 2.rectangle \n 3.triangle")
    ch=int(input("enter ypur choice"))
    if ch==1:
       a=int(input("Enter a side"))
       print("area=",sq(a))
    elif ch==2:
        l=int(input("enter length"))
        b=int(input("enter breadth"))
        print("area=",re(l,b))
    elif ch==3:
        l=int(input("enter base"))
        h=int(input("enter heigth"))
        print("area= ",tri(l,h))
    elif ch==4:
        break;
    else:
        print("Invalid choice")
```
